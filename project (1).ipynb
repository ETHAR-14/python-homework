1. Loading the Dataset

2. Data Exploration

This section explores the dataset by checking:
- Number of rows and columns
- Data types
- Missing values

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

# Search for the file in the current directory or subdirectories
file_name = 'bank_customers_time.xlsx'
file_path = None

for root, dirs, files in os.walk('.'):
    if file_name in files:
        file_path = os.path.join(root, file_name)
        break

if file_path:
    df = pd.read_excel(file_path, engine='openpyxl')
    print("Data loaded successfully")
    print("\n--- First 10 Rows ---")
    display(df.head(10))
    print("\n--- Last 10 Rows ---")
    display(df.tail(10))
else:
    print("Error: File not found. Please ensure the excel file is uploaded.")
```

3. Data Cleaning

The dataset was checked for missing values.
Missing numerical values were handled using mean replacement to ensure accurate analysis.

```python
print("Dataset Shape:")
print(df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())
```

4. Statistical Analysis

In this section, statistical calculations are performed using Pandas and NumPy, including:
- Mean
- Median
- Standard Deviation
- Minimum and Maximum values

```python
df.fillna(df.mean(numeric_only=True), inplace=True)

print("Missing values after cleaning:")
print(df.isnull().sum())
```

5. ## Histogram

The histogram shows the distribution of customer ages in the dataset.

```python
numeric_columns = ['Age', 'Balance', 'Income', 'Loan_Amount']

print("Mean:")
print(df[numeric_columns].mean())

print("\nMedian:")
print(df[numeric_columns].median())

print("\nStandard Deviation:")
print(df[numeric_columns].std())

print("\nMinimum Values:")
print(df[numeric_columns].min())

print("\nMaximum Values:")
print(df[numeric_columns].max())
```

6. Bar Chart

The bar chart compares the average income by gender.

```python
plt.figure(figsize=(8,5))

plt.hist(df['Age'], bins=10)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

7. Line Chart

The line chart shows the relationship between age and balance.

```python
avg_income = df.groupby('Gender')['Income'].mean()

plt.figure(figsize=(6,5))

plt.bar(avg_income.index, avg_income.values)

plt.title("Average Income by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Income")

plt.show()
```

8.  Insights

After analyzing the dataset, several patterns were observed:

- Income tends to increase for some age groups.
- There are differences in average income between genders.
- Most customers are concentrated within a specific age range.
- The dataset became more reliable after handling missing values.

Overall, data visualization helped identify relationships and trends clearly.

```python
sorted_df = df.sort_values(by='Age')

plt.figure(figsize=(8,5))

plt.plot(sorted_df['Age'], sorted_df['Balance'])

plt.title("Age vs Balance")
plt.xlabel("Age")
plt.ylabel("Balance")

plt.show()
```
