```python
import pandas as pd
import matplotlib.pyplot as plt

# Upload CSV file
  
df = pd.read_csv('bank_customers_dataset.csv')
df = pd.read_csv('bank_customers_dataset.csv')
# Read dataset
df = pd.read_csv('bank_customers_dataset.csv')

# Display first rows
print(df.head(10))

# Dataset information
print(df.tail(10))
print(df["balance"].mean())

print(df["balance"].median())

print(df["balance"].std())

print(df["balance"].min())

print(df["balance"].max())
# Missing values
print(df.isnull().sum())

# Remove duplicates
df = df.drop_duplicates()

# Statistical summary
print(df.describe())

# Average income by gender
avg_income = df.groupby('Gender')['Income'].mean()
print(avg_income)

avg_income.plot(kind='bar')
plt.xlabel('Gender')
plt.ylabel('Average Income')
plt.title('Average Income by Gender')
plt.show()

# Average credit score by gender
avg_credit = df.groupby('Gender')['Credit_Score'].mean()
print(avg_credit)

avg_credit.plot(kind='bar')
plt.xlabel('Gender')
plt.ylabel('Average Credit Score')
plt.title('Average Credit Score by Gender')
plt.show()

# Save cleaned dataset
df.to_csv('cleaned_bank_customers_dataset.csv', index=False)

print("Cleaned dataset saved successfully.")
```
